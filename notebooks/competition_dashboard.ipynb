```python
from pathlib import Path
import json, os, sys
ROOT = Path.cwd()
if not (ROOT / 'src' / 'orbitwin').exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / 'src'))
print('project:', ROOT)
```

```python
from orbitwin.config import load_config
from orbitwin.gmat_parser import parse_gmat_script
SCRIPT = ROOT / 'examples' / 'synthetic' / 'DifferentPlane_Circular.script'
CONFIG = ROOT / 'config' / 'quick.yaml'
config = load_config(CONFIG)
scenario, _ = parse_gmat_script(SCRIPT, config)
scenario.to_dict()
```

```python
from orbitwin.pipeline import solve_scenario
OUT = ROOT / 'output' / 'notebook-demo'
summary = solve_scenario(SCRIPT, config, OUT, render_top=10)
summary
```

```python
candidates = json.loads((OUT / 'candidates.json').read_text())
[(c['candidate_id'], c['total_dv_km_s'], c['intercept_time_s'], c['verified_by_gmat']) for c in candidates[:10]]
```
